```python
import cvxpy as cp
import numpy as np
```

```python
n = 3
x = cp.Variable(n,nonneg=True, name="x")
y = cp.Variable(n, integer=True, name="y")

c_x = np.array([2.0, 3.0, 1.5])
c_y = np.array([4.0, 2.5, 5.0])
```

```python
obj = cp.Maximize(c_x @ x + c_y @ y)
cons = [x + y >= np.array([5.5, 4.2, 6.0]), cp.sum(x) <= 15, y <= 10]
```

```python
prob = cp.Problem(obj, cons)
prob.solve()
```
