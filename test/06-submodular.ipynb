```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
def normalize_priors(priors):
    if np.sum(priors) == 0:
        return np.zeros_like(priors)
    return priors / np.sum(priors)
```

```python
def manipulation_thresholds(thresholds, priors, c):
    if np.sum(priors) == 0:
        return thresholds
    thresholds_m = np.maximum(0, thresholds - (normalize_priors(priors) / c))

    return thresholds_m
```

```python
def accuracy_loss(thresholds, priors, threshold_true, c):
    x_manipulation = manipulation_thresholds(thresholds, priors, c)
    losses = []
    for i in range(len(thresholds)):
        if thresholds[i] < threshold_true:
            loss = (threshold_true - x_manipulation[i])
        else:
            loss = np.abs(x_manipulation[i] - threshold_true)
        losses.append(loss)
    loss = np.array(losses)
    posteriors = normalize_priors(priors)
    return np.dot(losses, posteriors)
```

```python
def accuracy_loss_conditional(thresholds, priors, threshold_true, c):
    x_manipulation = manipulation_thresholds(thresholds, priors, c)
    losses = []
    for i in range(len(thresholds)):
        if thresholds[i] < threshold_true:
            loss = (threshold_true - x_manipulation[i])
        else:
            loss = np.abs(x_manipulation[i] - threshold_true)
        losses.append(loss)
    loss = np.array(losses)
    return np.dot(loss, priors)
```

```python
c = 20
threshold_true = 0.5
thresholds = np.array([0.1, 0.5, 0.9])

priors = np.array([1/3, 1/3, 1/3])

P = [0]
P3 = [0,2]

S = [0,1]
S3 = [0,1,2]

loss_P = accuracy_loss_conditional(thresholds[P], priors[P], threshold_true, c) / np.sum(priors[P3])
loss_P3 = accuracy_loss_conditional(thresholds[P3], priors[P3], threshold_true, c)

loss_S = accuracy_loss_conditional(thresholds[S], priors[S], threshold_true, c) / np.sum(priors[S3])
loss_S3 = accuracy_loss_conditional(thresholds[S3], priors[S3], threshold_true, c)
```

```python
loss_P3 - loss_P >= loss_S3 - loss_S
```
